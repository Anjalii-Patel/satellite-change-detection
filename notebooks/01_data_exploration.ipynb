# OSCD Data Exploration
This notebook verifies the co-registration of `imgs_1_rect` and `imgs_2_rect` and overlays the change labels on the images to confirm alignment.

```python
import matplotlib.pyplot as plt
import numpy as np
from scd.data import load_scene, get_splits
```

```python
def extract_rgb(img_array):
    # Extract RGB bands (indices 2, 1, 0 if bands are B02, B03, B04, B08)
    # B04 is Red, B03 is Green, B02 is Blue
    # data.py loads B02, B03, B04, B08
    rgb = img_array[[2, 1, 0], :, :]
    # Normalize for visualization (percentile clip)
    rgb_vis = np.zeros_like(rgb, dtype=np.float32)
    for i in range(3):
        p2, p98 = np.percentile(rgb[i], (2, 98))
        rgb_vis[i] = np.clip((rgb[i] - p2) / (p98 - p2), 0, 1)
    return np.transpose(rgb_vis, (1, 2, 0))

def plot_overlay(scene_id):
    t1, t2, label = load_scene(scene_id, split='train')
    
    rgb1 = extract_rgb(t1)
    rgb2 = extract_rgb(t2)
    
    fig, axs = plt.subplots(1, 3, figsize=(18, 6))
    axs[0].imshow(rgb1)
    axs[0].set_title(f'{scene_id} - T1')
    axs[0].axis('off')
    
    axs[1].imshow(rgb2)
    axs[1].set_title(f'{scene_id} - T2')
    axs[1].axis('off')
    
    # Overlay label on T2
    axs[2].imshow(rgb2)
    axs[2].imshow(label, cmap='Reds', alpha=0.5, interpolation='none')
    axs[2].set_title(f'{scene_id} - T2 with Change Mask Overlay')
    axs[2].axis('off')
    
    plt.tight_layout()
    plt.show()
```

```python
splits = get_splits()
print('Train scenes:', splits['train'])
print('Val scenes:', splits['val'])
```

```python
plot_overlay('bordeaux')
```

```python
plot_overlay('abudhabi')
```
